```python
def cac(rad):
    return 3.14*rad**2, 2*3.14*rad
ra = 10
area, cir = cac(ra)
print('반지름 10인 원의 면적은 ', area, '원의 둘레는 ', cir)
```

```python
def multi(n,m):
    s=[]
    for i in range(m):
        r = n*(i+1)
        s.append(r)
    return s
r1, r2, r3, r4 = multi(3,4)
print(r1, r2, r3, r4)
```
