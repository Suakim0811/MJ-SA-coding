```python
import requests
import json 

def translate(text, target = 'ko'):
    url = "https://translate.google.com/translate_a/single"
    params = {"client":"gtx", "sl":"auto", "tl": target, "dt": "t", "q": text}
    result = requests.get(url, params=params).json()
    return result[0][0][0]

txt = input('origin text: ')
tg_lang = input('target lenguage: ')
print('translated txt: ', end = '')
print(translate(txt, tg_lang))
```

```python
import requests

def weather():
    url = "https://www.weather.go.kr/w/renew2021/rest/main/daily-forecast-stn.do"
    params = {
        "stns": "143"
    }
    res = requests.get(url, params=params).json()
    return res['data'][0]['forecast'][0]['wf']

print(weather())
```

```python
import requests
from bs4 import BeautifulSoup
url = "https://quotes.toscrape.com/"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

qoute_block = soup.find_all('div', class_='quote')

for i in qoute_block:
    qoute = i.find('span', class_='text').text
    author = i.find('small', class_='author').text
    print(qoute, '-', author)
    
```

```python
import requests
from bs4 import BeautifulSoup
url = 'http://books.toscrape.com/'
response = requests.get(url)
response.encoding = 'utf-8' 
soup = BeautifulSoup(response.text, 'html.parser')

book_block = soup.find_all('article', class_='product_pod')


for i in book_block:
    name = i.find('h3').find('a').get('title')
    price = i.find('p', class_='price_color').text
    instock = i.find('p', class_='instock availability').text
    print(f'책 제목 : \033[34m{name}\33[0m, 가격: \033[33m{price}\033[0m, 재고상태: \033[32m{instock.strip()}\33[0m')
    
```
