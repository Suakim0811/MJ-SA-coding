```python
# 9295
c = list(map(int, input().split()))
a = []
b = []
for n in range(len(c)):
    if n == 0:
        continue
    if n%2 == 0:
        a.append(c[n])
    if n%2 != 0:
        b.append(c[n])
for i in range(len(a)):
    print(f'case{i+1} : {a[i]+b[i]}')
    
```

```python
# 10101
a = list(map(int, input().split()))
for i in range(3):
    if a[i] + a[i-1] == 120 == 2*a[i] and a[i]==60:
        print('Equilateral = 정삼각형')
        break
    if a[i]+a[i-1]+a[i+1<3] == 180:
        if 180 - a[i] == 2*a[i-1] == 2*a[i+1<3]:
            print('Isosceles = 이등변')
            break
        if a[i] != a[i-1] != a[i+1<3]:
            print('Scalene = 그냥 삼각형')
            break
    else:
        print('Error')
        break

```
