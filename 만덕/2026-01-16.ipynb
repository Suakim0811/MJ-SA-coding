```python
# 2444
num = int(input())
for i in range(num):
    s = '*'*(2*i + 1)
    b = ' '*(num - i -1)
    print(b+s)
for u in range(num-1, -1, -1):
    s = '*'*(2*(u-1)+1)
    b = ' '*(num - u)
    print(b+s)
```
