```python
import random
def estimate_pi(num_samples : int) :
    cnt = 0
    for i in range(num_samples):
        x = random.random()
        y = random.random()
        if x**2+y**2<=1:
            cnt += 1
    print(4*cnt/num_samples)

estimate_pi(num_samples=50000000)

    
```

```python
find = int(input())
num = [1,3,5,7,9,10,11,13,16,17,18,20,21]
left = 0
right = len(num)-1
mid = (left+right)//2
is_found = 0

while left < right :
    if find > num[mid]:
        left = mid + 1
    elif find < num[mid]:
        right = mid - 1
    else:
        is_found = 1
        break
    mid = (left+right)//2

if is_found:
    print('location:', mid)
else:
    print('not found')
```

```python
i = int(input())
res = 0
cnt = 0
while i:
    if i > 0:
        res += i
    else:
        cnt += 1
    i = int(input())
print(f'양수 합 : {res} 음수 개수 : {cnt}')
```

```python
i = int(input())
pcnt = 0
psum = 0
pmax = 0
nmin = 0
while i:
    if i>0:
        psum += i
        pcnt += 1
        if i > pmax:
            pmax = i
    elif nmin > i:
        nmin = i
    i = int(input())
print(f'합꼐 : {psum} 평균 : {psum/pcnt} 양수 최대 : {pmax} 음수 최소 : {nmin}')
```
